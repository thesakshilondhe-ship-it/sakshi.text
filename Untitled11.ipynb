```python
import csv

# Create a dummy medical_data.csv file for demonstration
# In a real scenario, this file would already exist or be generated elsewhere.
medical_data_content = [
    {"question": "What is the common cold?", "answer": "The common cold is a viral infection of your nose and throat. It's usually harmless, although it might not feel that way."},
    {"question": "How to treat a fever?", "answer": "Fevers usually go away on their own within a few days. You can take acetaminophen (Tylenol, others) or ibuprofen (Advil, Motrin IB, others) to help reduce a fever."},
    {"question": "What are symptoms of flu?", "answer": "Flu symptoms include fever, cough, sore throat, runny or stuffy nose, muscle or body aches, headaches, and fatigue."}
]

with open("medical_data.csv", "w", newline="") as file:
    fieldnames = ["question", "answer"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(medical_data_content)

data = []

with open("medical_data.csv", "r") as file: # Corrected filename from .py to .csv
    reader = csv.DictReader(file) # Corrected from py.DictReader to csv.DictReader
    for row in reader:
        data.append(row)

print("Medical Chatbot Started!")

question = input("Ask a question: ")

found = False

for row in data:
    if question.lower() == row["question"].lower():
        print("Answer:", row["answer"])
        found = True

if not found:
    print("Sorry, answer not found.")
```
